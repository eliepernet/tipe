```python
import matplotlib.pyplot as plt
import numpy as np
import pyvista as pv
```

```python
mesh = pv.read('../data/Solid_77.vtk')

# Définir un plan de coupe (par un point et une normale)
slice_plane = mesh.slice(normal='z')  # Normal à l'axe z (par défaut au centre du mesh)

# Visualiser
plotter = pv.Plotter()
plotter.add_mesh(slice_plane, color='red')  # tranche
plotter.add_mesh(mesh, color='lightgray', opacity=0.3)  # modèle 3D en fond
plotter.show()
```

```python
edges = slice_plane.extract_feature_edges(boundary_edges=True)
contour_points = np.array(edges.points)  # tableau Nx3
# contour_points = np.delete(contour_points, np.random.choice(range(len(contour_points)), 1300, replace=False), 0)
plt.scatter(contour_points[:, 0],  contour_points[:, 1])
plt.title("Points du contour")
plt.axis('equal')
plt.show()
```
